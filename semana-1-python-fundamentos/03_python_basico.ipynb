```python
# Función 1: calcula el total de una venta
def calcular_total(precio, cantidad):
    total = precio * cantidad
    return total

# Función 2: aplica un descuento porcentual a un monto
def aplicar_descuento(monto, porcentaje):
    descuento = monto * (porcentaje / 100)
    total_con_descuento = monto - descuento
    return total_con_descuento
```

```python
# Probar calcular_total
venta1 = calcular_total(250000, 3)
print(venta1) 
```

```python
venta2 = calcular_total(180000, 1)
print(venta2) 
```

```python
venta3 = calcular_total(340000,10)
print(venta3)
```

```python
# Probar aplicar_descuento
precio_final = aplicar_descuento(750000, 10)
print(precio_final)
```

```python
precio_final = aplicar_descuento(340000, 20)
print(precio_final)
```

```python
total_final = aplicar_descuento(calcular_total(250000, 3), 15)
print(total_final)
```
