```python
from faker import Faker
import pandas as pd
import random

fake = Faker("es_CO")
```

```python
productos = ["Persiana enrollable", "Cortina blackout", "Persiana veneciana", "Cortina translucida", "Panel japones"]
colores = ["Blanco", "Gris", "Beige", "Negro", "Azul"]
vendedores = ["Ana Torres", "Luis Ramirez", "Sofia Gomez"]
```

```python
productos
```

```python
registros = []
for i in range(200):
    registros.append({
        "id_venta": i + 1,
        "cliente": fake.name(),
        "producto": random.choice(productos),
        "color": random.choice(colores),
        "vendedor": random.choice(vendedores),
        "monto": random.randint(80000, 500000),
        "fecha": fake.date_between(start_date="-6M", end_date="today")
    })

df = pd.DataFrame(registros)
df.head(15)
```

```python
df.shape
```

```python
import numpy as np

# Vuelve nulos algunos clientes al azar
indices_nulos = random.sample(range(len(df)), 8)
df.loc[indices_nulos, "cliente"] = None

# Duplica 5 filas a propósito
duplicados = df.sample(5)
df = pd.concat([df, duplicados], ignore_index=True)

df.shape
```

```python
df.isnull().sum()
```

```python
df.to_csv("../data/ventas_pentagrama_sim.csv", index=False)
```
